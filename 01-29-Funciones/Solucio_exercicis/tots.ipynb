 Crea una funció que rebi un *nombre enter* i **retorni el seu doble**.


```python
def doble(nombre):
    return nombre * 2

# Exemple d'ús
numero = 10
resultat = doble(numero)
print("El doble de ",  numero, " es ", resultat)
```

## Funció amb un enter (càlcul senzill)

Crea una funció que rebi un nombre enter i retorni `True` si és **parell** o `False` si és **senar`.

Un nombre és divisible per 2 quan el seu residu és 0.


```python
def esParell(nombre):
    return nombre % 2 == 0

# Exemples d'ús
print(" 4 es parell?", esParell(4))  # True
print(" 7 es parell?", esParell(7))  # False
```

## Funció amb una llista (càlcul)

Crea una funció que rebi una llista de nombres (`numeros`) i retorni la **suma de tots els elements que superin el valor 40**.

A la llista hi ha d’haver nombres **més grans i més petits de 40**.

```python
def sumaGrans40(numeros):
    suma = 0
    for numUnic in numeros:
        if numUnic > 40:
            suma += numUnic
    return suma

# Exemples d'ús
llista = [10, 45, 6, 130, 80, 25]
print("La suma dels nombres superiors a 40 és:", sumaGrans40(llista)) 
```

## Funció amb una llista

Crea una funció que rebi una llista de nombres (`numeros`) i retorni el **nombre més gran** de la llista.

Hauràs de **recórrer tota la llista** i guardar el valor més gran en una variable.  
En cada volta, aquesta variable pot canviar si es troba un nombre més gran.


```python
def mesGranLlista(numeros):
    maxim = numeros[0]
    for n in numeros:
        if n > maxim:
            maxim = n
    return maxim

# Exemples d'ús
llista = [12, 45, 3, 89, 27]
print("El nombre més gran de la llista és:", llista)
print(mesGranLlista(llista))
```
