Crea una classe Llum amb l'atribut ubicacio amb un atribut estat (booleà, comença en False). Afegeix mètodes per encendre i apagar. I un mètode que et digui siEstaEncesa, escrivint una frase. Al programa principal crea dos llums diferents, apagar una i encen l’altra i pregunta si estanEnceses les dues.
, fes-ho a exercicisPoo.py i el programa principal per provar també, no borris el que hi ha

```python
class Llum:
    def __init__(self, ubicacio):
        self.ubicacio = ubicacio
        self.estat = False

    def encendre(self):
        self.estat = True

    def apagar(self):
        self.estat = False

    def siEstaEncesa(self):
        if self.estat:
            print(f"La llum de {self.ubicacio} està encesa.")
        else:
            print(f"La llum de {self.ubicacio} està apagada.")


if __name__ == "__main__":
    llum1 = Llum("cuina")
    llum2 = Llum("menjador")
    
    llum1.encendre()
    llum1.apagar()  # Ja està apagada, però per seguir les instruccions
    llum2.encendre()
    
    llum1.siEstaEncesa()
    llum2.siEstaEncesa()
```

Crea una classe Gos amb l'atribut nom amb un atribut fam (valor inicial 10). Crea un mètode menjar que resti 2 a la fam (escrigui estic menjant també). Crea un segon mètode que és digui menjarQuantitat, que de parametre d'entrada apart del shelf, rebi Quantitat, i resti aquesta quantitat a la fam. Recordeu que mai pot quedar amb negatiu, la fam, si és negatiu, deixar-la a 0.
Crea un mètode que és digui acariciar, i et digui/escrigui si et mossega, si té més de 4 de fam, o si mou la cua. Fes el programa per provar quan et mossega i quan no.

```python
class Gos:
    def __init__(self, nom):
        self.nom = nom
        self.fam = 10

    def menjar(self):
        self.fam -= 2
        if self.fam < 0:
            self.fam = 0
        print("Estic menjant")

    def menjarQuantitat(self, quantitat):
        self.fam -= quantitat
        if self.fam < 0:
            self.fam = 0

    def acariciar(self):
        if self.fam > 4:
            print(f"El gos {self.nom} et mossega!")
        else:
            print(f"El gos {self.nom} mou la cua.")


# Programa per provar
gos = Gos("Yoska")

print(f"Fam inicial: {gos.fam}")
gos.acariciar()  # Ha de moure la cua, fam=10 >4

gos.menjar()  # Resta 2, fam=8
print(f"Fam després de menjar: {gos.fam}")
gos.acariciar()  # Encara mou la cua, fam=8 >4

gos.menjarQuantitat(5)  
print(f"Fam després de menjar quantitat 5: {gos.fam}")
gos.acariciar()  

gos.menjarQuantitat(10)  
print(f"Fam després de menjar quantitat 10: {gos.fam}")
gos.acariciar()  # Mou la cua
```

Crea una classe Compte amb titular i saldo. Afegeix mètodes per ingressar (sumar) i retirar diners. Alerta: no es pot retirar més del que hi ha! Si ingresses mes d'un milio d'euros, ha de mostrar una frase que digui, aviso a la policia perquè verifiqui que aquest diner és legal (però igualment ingressa els diners). Fes un programa que provi aquest mètodes.

```python
class Compte:
    def __init__(self, titular, saldo=0):
        self.titular = titular
        self.saldo = saldo

    def ingressar(self, quantitat):
        self.saldo += quantitat
        if quantitat > 1000000:
            print("Aviso a la policia perquè verifiqui que aquest diner és legal.")
        print(f"S'han ingressat {quantitat} euros. Saldo actual: {self.saldo} euros.")

    def retirar(self, quantitat):
        if quantitat > self.saldo:
            print(f"No es pot retirar {quantitat} euros. Saldo insuficient. Saldo actual: {self.saldo} euros.")
        else:
            self.saldo -= quantitat
            print(f"S'han retirat {quantitat} euros. Saldo actual: {self.saldo} euros.")


# Programa per provar
compte = Compte("Joan", 500)

print(f"Compte de {compte.titular} amb saldo inicial: {compte.saldo} euros.")

compte.ingressar(200)  # Ingressar menys d'un milió
compte.retirar(100)    # Retirar
compte.retirar(700)    # Intentar retirar més del que hi ha
compte.ingressar(1500000)  # Ingressar més d'un milió
```

Crea una classe Producte amb nom, preu i estoc. Afegeix un mètode vendre(quantitat) que redueixi l'estoc només si n'hi ha prou.

```python
class Producte:
    def __init__(self, nom, preu, estoc):
        self.nom = nom
        self.preu = preu
        self.estoc = estoc

    def vendre(self, quantitat):
        if quantitat <= self.estoc:
            self.estoc -= quantitat
            print(f"S'han venut {quantitat} unitats de {self.nom}. Estoc restant: {self.estoc}.")
        else:
            print(f"No hi ha prou estoc de {self.nom}. Estoc disponible: {self.estoc}.")


# Programa per provar
producte = Producte("Llibre", 15.99, 10)

print(f"Producte: {producte.nom}, Preu: {producte.preu}€, Estoc: {producte.estoc}")

producte.vendre(3)  # Vendre 3, n'hi ha prou
producte.vendre(8)  # Intentar vendre 8, no n'hi ha prou
```

Crea una classe Tasca amb titol i un atribut feta (inicialment False). Afegeix un mètode completar.

```python
class Tasca:
    def __init__(self, titol):
        self.titol = titol
        self.feta = False

    def completar(self):
        self.feta = True
        print(f"La tasca '{self.titol}' s'ha completat.")


# Programa per provar
tasca = Tasca("Comprar pa")

print(f"Tasca: {tasca.titol}, Feta: {tasca.feta}")

tasca.completar()

print(f"Després de completar {tasca.titol}: Feta: {tasca.feta}")
```

Crea una classe Termostat amb una temperatura actual. Afegeix mètodes per apujar i abaixar la temperatura, però amb un límit màxim de 30°C i mínim de 15°C.

```python
class Termostat:
    def __init__(self, temperatura_actual=20):
        self.temperatura_actual = temperatura_actual

    def apujar(self, graus=1):
        nova_temp = self.temperatura_actual + graus
        if nova_temp > 30:
            print("No es pot apujar més. Límit màxim 30°C.")
            self.temperatura_actual = 30
        else:
            self.temperatura_actual = nova_temp
        print(f"Temperatura actual: {self.temperatura_actual}°C")

    def abaixar(self, graus=1):
        nova_temp = self.temperatura_actual - graus
        if nova_temp < 15:
            print("No es pot abaixar més. Límit mínim 15°C.")
            self.temperatura_actual = 15
        else:
            self.temperatura_actual = nova_temp
        print(f"Temperatura actual: {self.temperatura_actual}°C")


# Programa per provar
termostat = Termostat(22)

print(f"Temperatura inicial: {termostat.temperatura_actual}°C")

termostat.apujar(5)  # Apujar 5 graus
termostat.apujar(10)  # Intentar apujar 10, però supera el límit
termostat.abaixar(20)  # Intentar abaixar 20, però supera el límit
termostat.abaixar(5)  # Abaixar 5 graus
```

Crea una classe Cercle que rebi el radi. Crea mètodes per calcular el perímetre i l'àrea utilitzant la llibreria math.
Has de posar al principi import math, i quant vulguis saber el número pi, és math.pi

```python
import math

class Cercle:
    def __init__(self, radi):
        self.radi = radi

    def calcular_perimetre(self):
        return 2 * math.pi * self.radi

    def calcular_area(self):
        return math.pi * self.radi ** 2


# Programa per provar
cercle = Cercle(5)

print(f"Cercle amb radi {cercle.radi}")
print(f"Perímetre: {cercle.calcular_perimetre():.2f}")
print(f"Àrea: {cercle.calcular_area():.2f}")
```
